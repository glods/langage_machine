## 1. Le Probleme : Pas de notion de similarite

### 1.1 Representation One-Hot (une representation simple des mots en vecteurs )

```python
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np

# Vocabulaire simple
vocab = ["roi", "reine", "homme", "femme", "excellent", "genial"]

# Representation One-Hot
print("=== Representation One-Hot ===")
print("Chaque mot = un vecteur avec UN SEUL 1\n")

for i, mot in enumerate(vocab):
    vecteur = [0] * len(vocab)
    vecteur[i] = 1
    print(f"{mot:12} = {vecteur}")

print("\n[!] Tous les mots sont a la MEME distance les uns des autres !")
```

```python
from scipy.spatial.distance import cosine

# Calculons les distances
def one_hot(mot, vocab):
    v = [0] * len(vocab)
    v[vocab.index(mot)] = 1
    return v

v_roi = one_hot("roi", vocab)
v_reine = one_hot("reine", vocab)
v_excellent = one_hot("excellent", vocab)
v_genial = one_hot("genial", vocab)

print("=== Distances One-Hot (cosine) ===")
print(f"Distance roi-reine     : {cosine(v_roi, v_reine):.2f}")
print(f"Distance excellent-genial : {cosine(v_excellent, v_genial):.2f}")
print(f"Distance roi-excellent : {cosine(v_roi, v_excellent):.2f}")
print("\n[!] Toutes les distances = 1.0 (distance maximale)")
print("    'excellent' n'est PAS plus proche de 'genial' que de 'roi'")
```

### 1.2 Ce qu'on voudrait

On voudrait que :
- "roi" soit proche de "reine" (meme champ semantique)
- "excellent" soit proche de "genial" (synonymes)
- "roi" soit loin de "excellent" (pas de relation)

## 2. L'idee : Word Embeddings

### 2.1 Representation dense

Au lieu d'un vecteur sparse (un seul 1), on utilise un vecteur **dense** ou chaque dimension capture un aspect du sens.

```python
# Exemple simplifie (4 dimensions)
# En realite : 100-300 dimensions

# Dimensions imaginaires : [Royaute, Genre(M+/F-), Age, Pouvoir]
embeddings_exemple = {
    "roi":    [ 0.9,  0.8,  0.5,  0.9],
    "reine":  [ 0.9, -0.8,  0.5,  0.9],
    "homme":  [ 0.1,  0.8,  0.5,  0.2],
    "femme":  [ 0.1, -0.8,  0.5,  0.2],
    "prince": [ 0.8,  0.7, -0.7,  0.4],
}

print("=== Representation Dense (Word Embeddings) ===")
print("Dimensions : [Royaute, Genre, Age, Pouvoir]\n")

for mot, vec in embeddings_exemple.items():
    print(f"{mot:8} = {vec}")
```

```python
# Maintenant les distances ont du sens !
print("=== Distances avec Embeddings ===")
print(f"Distance roi-reine   : {cosine(embeddings_exemple['roi'], embeddings_exemple['reine']):.3f}")
print(f"Distance roi-homme   : {cosine(embeddings_exemple['roi'], embeddings_exemple['homme']):.3f}")
print(f"Distance homme-femme : {cosine(embeddings_exemple['homme'], embeddings_exemple['femme']):.3f}")
print("\n[OK] roi est PLUS PROCHE de reine que de homme !")
```

## 3. L'arithmetique des mots

### 3.1 L'equation magique

```python
import numpy as np

# roi - homme + femme = ?
v_roi = np.array(embeddings_exemple["roi"])
v_homme = np.array(embeddings_exemple["homme"])
v_femme = np.array(embeddings_exemple["femme"])
v_reine = np.array(embeddings_exemple["reine"])

print("=== L'equation magique ===")
print(f"roi     = {list(v_roi)}")
print(f"- homme = {list(v_homme)}")
print(f"+ femme = {list(v_femme)}")
print("-" * 40)

resultat = v_roi - v_homme + v_femme
print(f"        = {list(resultat)}")
print(f"\nreine   = {list(v_reine)}")
print("\n[OK] Le resultat est tres proche de 'reine' !")
```

```python
# Verification : quel mot est le plus proche du resultat ?
def trouver_plus_proche(vecteur_cible, embeddings):
    distances = {}
    for mot, vec in embeddings.items():
        distances[mot] = cosine(vecteur_cible, vec)
    return sorted(distances.items(), key=lambda x: x[1])

print("=== Mots les plus proches de (roi - homme + femme) ===")
for mot, dist in trouver_plus_proche(resultat, embeddings_exemple):
    print(f"{mot:8} : distance = {dist:.4f}")
```

## 4. Utilisation de vrais Word Embeddings

### 4.1 Avec Gensim et Word2Vec pre-entraine

```python
# Installation si necessaire
# !pip install gensim

# Pour cet exemple, on utilise un petit modele
# En production, utilisez des modeles pre-entraines complets

from gensim.models import Word2Vec
import warnings
warnings.filterwarnings('ignore')

# Corpus d'entrainement (simplifie)
corpus = [
    ["le", "roi","est", "un", "homme", "qui", "gouverne", "le", "royaume" "et"],
    ["la", "reine", "gouverne", "le", "pays"],
    ["le", "roi", "et", "la", "reine", "sont", "au", "chateau"],
    ["homme", "travaille", "dur"],
    ["femme", "travaille", "dur"],
    ["le", "prince", "est", "jeune"],
    ["la", "princesse", "est", "jeune"],
    ["ce", "cours", "est", "excellent"],
    ["cette", "formation", "est", "geniale"],
    ["le", "cours", "est", "genial"],
    ["formation", "excellente", "je", "recommande"],
]

# Entrainement Word2Vec
model = Word2Vec(
    sentences=corpus,
    vector_size=50,  # Dimensions
    window=3,        # Contexte
    min_count=1,     # Frequence min
    epochs=100       # Iterations
)

print("[OK] Modele Word2Vec entraine")
print(f"    Vocabulaire : {len(model.wv)} mots")
print(f"    Dimensions  : {model.wv.vector_size}")
```

```python
# Voir le vecteur d'un mot
print("=== Vecteur du mot 'roi' (10 premieres dimensions) ===")
print(model.wv["roi"][:10])
```

```python
# Mots similaires
print("=== Mots similaires a 'roi' ===")
try:
    for mot, score in model.wv.most_similar("roi", topn=5):
        print(f"  {mot:15} : {score:.3f}")
except KeyError as e:
    print(f"Mot non trouve: {e}")
```

```python
# Similarite entre deux mots
print("=== Similarite entre mots ===")
paires = [("roi", "reine"), ("cours", "formation"), ("roi", "cours")]
for m1, m2 in paires:
    try:
        sim = model.wv.similarity(m1, m2)
        print(f"  {m1:12} - {m2:12} : {sim:.3f}")
    except KeyError as e:
        print(f"  Mot non trouve: {e}")
```

### 4.2 Avec SpaCy (plus simple pour le francais)

```python
# Installation si necessaire
# !pip install spacy
# !python -m spacy download fr_core_news_md

try:
    import spacy
    nlp = spacy.load("fr_core_news_md")
    
    print("=== SpaCy avec modele francais ===")
    
    # Similarite
    mots = ["roi", "reine", "homme", "femme", "voiture"]
    docs = [nlp(mot) for mot in mots]
    
    print("\nSimilarites :")
    print(f"  roi-reine   : {docs[0].similarity(docs[1]):.3f}")
    print(f"  roi-homme   : {docs[0].similarity(docs[2]):.3f}")
    print(f"  roi-voiture : {docs[0].similarity(docs[4]):.3f}")
    
except OSError:
    print("[!] Modele SpaCy non installe")
    print("    Pour installer : python -m spacy download fr_core_news_md")
```

## 5. Visualisation des embeddings

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Recuperer les vecteurs
mots = list(model.wv.key_to_index.keys())
vecteurs = [model.wv[mot] for mot in mots]

# Reduction a 2D avec PCA
pca = PCA(n_components=2)
vecteurs_2d = pca.fit_transform(vecteurs)

# Visualisation
plt.figure(figsize=(12, 8))
plt.scatter(vecteurs_2d[:, 0], vecteurs_2d[:, 1], alpha=0.7, s=100, c='#10274a')

for i, mot in enumerate(mots):
    plt.annotate(mot, (vecteurs_2d[i, 0], vecteurs_2d[i, 1]), 
                 fontsize=10, ha='center', va='bottom')

plt.title('Word Embeddings projetes en 2D (PCA)', fontsize=14, fontweight='bold', color='#10274a')
plt.xlabel('Dimension 1')
plt.ylabel('Dimension 2')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('embeddings_2d.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()

print("\n[i] Les mots similaires sont regroupes dans l'espace !")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.manifold import TSNE

# Recuperer les vecteurs
mots = list(model.wv.key_to_index.keys())
vecteurs = np.array([model.wv[mot] for mot in mots])

# Ajustement automatique de la perplexity
n_samples = len(vecteurs)
perplexity_value = min(30, n_samples - 1)

# Reduction a 2D avec t-SNE
tsne = TSNE(
    n_components=2,
    perplexity=perplexity_value,
    learning_rate='auto',
    init='pca',
    random_state=42
)

vecteurs_2d = tsne.fit_transform(vecteurs)

# Visualisation
plt.figure(figsize=(12, 8))
plt.scatter(vecteurs_2d[:, 0], vecteurs_2d[:, 1], alpha=0.7, s=100, c='#10274a')

for i, mot in enumerate(mots):
    plt.annotate(
        mot,
        (vecteurs_2d[i, 0], vecteurs_2d[i, 1]),
        fontsize=10, ha='center', va='bottom'
    )

plt.title('Word Embeddings projetes en 2D (t-SNE)', fontsize=14, fontweight='bold', color='#10274a')
plt.xlabel('Dimension 1')
plt.ylabel('Dimension 2')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('embeddings_tsne_2d.png', dpi=150, bbox_inches='tight', facecolor='white')
plt.show()

print("\n[i] Les mots similaires sont regroupes dans l'espace (t-SNE) !")
```

## 6. Application : Classification avec Embeddings

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import numpy as np

# Corpus d'avis
avis = [
    ("cours excellent je recommande", 1),
    ("formation geniale tres bien", 1),
    ("super cours j'ai appris", 1),
    ("formation difficile incomprehensible", 0),
    ("cours confus mal explique", 0),
    ("decevant pas clair du tout", 0),
]

# Fonction pour obtenir l'embedding d'une phrase (moyenne des mots)
def phrase_to_vector(phrase, model):
    mots = phrase.lower().split()
    vecteurs = []
    for mot in mots:
        if mot in model.wv:
            vecteurs.append(model.wv[mot])
    if vecteurs:
        return np.mean(vecteurs, axis=0)
    else:
        return np.zeros(model.wv.vector_size)

# Preparer les donnees
X = np.array([phrase_to_vector(phrase, model) for phrase, _ in avis])
y = np.array([label for _, label in avis])

print("=== Classification avec Word Embeddings ===")
print(f"Forme des features : {X.shape}")
print(f"Labels : {y}")
```

```python
# Entrainement du classifieur
clf = LogisticRegression()
clf.fit(X, y)

# Test sur de nouveaux avis
nouveaux_avis = [
    "formation excellente",
    "cours tres bien fait",
    "difficile a suivre",
    "genial je recommande"
]

print("=== Predictions ===")
for avis in nouveaux_avis:
    vec = phrase_to_vector(avis, model).reshape(1, -1)
    pred = clf.predict(vec)[0]
    proba = clf.predict_proba(vec)[0]
    sentiment = "POSITIF" if pred == 1 else "NEGATIF"
    print(f"  '{avis}' --> {sentiment} (confiance: {max(proba)*100:.0f}%)")
```

## 7. Recapitulatif

### Ce que nous avons appris

| Concept | Description |
|---------|-------------|
| **Word Embedding** | Vecteur dense qui capture le SENS d'un mot |
| **One-Hot** | Representation sparse (un seul 1) - pas de semantique |
| **Word2Vec** | Algorithme pour apprendre les embeddings |
| **CBOW** | Predire le mot central depuis le contexte |
| **Skip-gram** | Predire le contexte depuis le mot central |

### Forces
- [+] Capture la semantique
- [+] Vecteurs denses (100-300 dim vs 10 000+)
- [+] Analogies : roi - homme + femme = reine
- [+] Modeles pre-entraines disponibles

### Limites
- [x] Un vecteur par mot (pas de contexte)
- [x] "avocat" fruit = "avocat" juriste
- [x] Mots hors vocabulaire

### La serie complete

| Acte | Technique | Ce qu'on apprend |
|------|-----------|------------------|
| I | Bag of Words | Transformer texte en vecteur |
| II | TF-IDF | Ponderer les mots importants |
| III | Naive Bayes | Classifier avec les probabilites |
| IV | N-grammes | Capturer l'ordre des mots |
| **V** | **Word Embeddings** | **Comprendre la semantique** |

## 8. Et apres ? Les Transformers

Les Word Embeddings donnent UN vecteur par mot, independamment du contexte.

Mais "avocat" dans "Je mange un avocat" n'a pas le meme sens que "avocat" dans "Mon avocat plaide".

Les **Transformers** (BERT, GPT) resolvent ce probleme en donnant un vecteur DIFFERENT selon le contexte !
